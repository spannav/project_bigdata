## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |
| **Aspect analysis** | คนพูดถึงราคา คุณภาพ การส่งของ ฯลฯ อย่างไร |
| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่


## แบบฟอร์ม Insight Card

ทุก insight เขียนในรูปแบบนี้ 5 บรรทัด ไม่ต้องยาว

> **Insight:** ประโยคเดียวที่บอกสิ่งที่พบ ต้องมีตัวเลขอยู่ในประโยค \
> **หลักฐาน:** ตัวเลขพร้อม n และชี้ว่าดูได้จากกราฟหรือตารางไหน \
> **So what:** กระทบธุรกิจอย่างไร ถ้าประมาณขนาดผลกระทบเป็นตัวเลขได้ยิ่งดี \
> **Action:** ทีมไหนควรทำอะไร และจะรู้ได้อย่างไรว่าได้ผล \
> **ความมั่นใจ:** สูง กลาง หรือ ต่ำ พร้อมเหตุผลสั้น ๆ

**ตัวอย่างที่ยังไม่ใช่ insight**
> ลูกค้าพูดถึงเรื่องราคาบ่อยที่สุด คิดเป็น 32% ของรีวิว

เป็นเพียงข้อสังเกต ไม่ได้บอกว่าใครต้องทำอะไร และ 32% นั้นมากหรือน้อยก็ไม่รู้เพราะไม่มีอะไรให้เทียบ

**ตัวอย่างที่ใช้ได้**
> **Insight:** รีวิวที่พูดถึงบรรจุภัณฑ์รั่วมีเพียง 6.2% ของทั้งหมด (n = 3,100) แต่ได้ดาวเฉลี่ย 2.1 เทียบกับค่าเฉลี่ยรวม 4.0 และกระจุกอยู่ในสินค้า 12 ชิ้นที่ใช้ขวดแบบปั๊ม \
> **หลักฐาน:** ตารางค่าเฉลี่ยดาวแยกตามด้าน และกราฟแท่งสินค้า 12 ชิ้นในผลวิเคราะห์ \
> **So what:** สินค้า 12 ชิ้นนี้คิดเป็น 18% ของยอดรีวิวทั้งหมวด ถ้าแก้ปัญหาฝาขวดได้ คาดว่าคะแนนเฉลี่ยของหมวดจะขยับขึ้นราว 0.15 ดาว \
> **Action:** ทีม Product ตรวจสอบซัพพลายเออร์ฝาขวดของสินค้า 12 รายการนี้ภายในไตรมาสหน้า แล้วติดตามสัดส่วนรีวิวที่มีคำว่า leak, spill, broken cap ซึ่งตอนนี้อยู่ที่ 14% ต่อเดือน \
> **ความมั่นใจ:** กลาง ยืนยันได้จากข้อความ แต่ยังไม่มีข้อมูลการคืนสินค้ามาช่วยยืนยัน

### 2.2 เตรียมข้อมูลที่เชื่อถือได้

ทำเพียงสิ่งจำเป็นก่อนเริ่มวิเคราะห์

* สุ่มรีวิวอย่างน้อย **20,000 รายการ** และอย่าใช้เพียง N บรรทัดแรก
* เชื่อม review กับ product metadata ด้วย `parent_asin`
* รายงานสั้น ๆ ว่า join สำเร็จกี่เปอร์เซ็นต์ และมี missing/duplicate สำคัญหรือไม่
* เขียน 2–3 บรรทัดว่า **ข้อมูลนี้เป็นตัวแทนของใคร และไม่เป็นตัวแทนของใคร**

> จุดประสงค์คือให้รู้ว่าข้อมูลที่ใช้เชื่อถือได้พอสำหรับคำถามของทีม ไม่ใช่ทำ data cleaning ทุกอย่างที่เป็นไปได้


````python
# สุ่มรีวิว 20,000 รายการจากไฟล์
df_review_sample = df_review.sample(n=20000, random_state=42).copy()
print(f"จำนวนรีวิวหลังจากสุ่มได้: {df_review_sample.shape[0]} รายการ")

# merge ข้อมูลรีวิวกับข้อมูล meta โดยใช้ parent_asin เป็น key
table_merge = df_review_sample.merge(
    df_meta,
    on="parent_asin",
    how="left",
    indicator=True
)
print(f"จำนวนรีวิวหลัง merge : {table_merge.shape[0]} รายการ")

# รายงานผลการ join
join_success = (
    (table_merge["_merge"] == "both").sum()
    / len(table_merge)
    * 100
)

print(f"Join success rate: {join_success:.2f}%")

#ตรวจค่าซ้ำซ้อนของรีวิว 
duplicate_count = table_merge.duplicated(
    subset=["user_id", "parent_asin", "timestamp", "text"]
).sum()

print("รีวิวซ้ำในตารางหลัง merge:", duplicate_count)
````

````python
def check_missing(df):
    # คอลัมน์ที่จำเป็นต่อ Brief B1
    important_cols = [
        'parent_asin',
        'rating',
        'text',
        'average_rating',
        'rating_number',
        'categories',
        'features'
    ]

    print("===== Missing Value Check =====")

    for col in important_cols:
        if col not in df.columns:
            print(f"ไม่พบคอลัมน์: {col}")
            continue

        missing_count = df[col].isna().sum()
        missing_percent = (missing_count / len(df)) * 100

        if missing_count > 0:
            print(
                f"{col}: มีค่า null {missing_count:,} ค่า "
                f"({missing_percent:.2f}%)"
            )
        else:
            print(f"{col}: ไม่มีค่า null")

print("ตรวจสอบตารางรีวิวหลัง merge:")
check_missing(table_merge)
````

````python
#สำคัญในแง่การหา category ที่น่าสนใจ

table_merge['low_rating'] = table_merge['rating'] <= 2

category_df = table_merge.explode('categories')

category_summary = (
    category_df
    .groupby('categories')
    .agg(
        total_reviews=('low_rating', 'size'),
        low_rating_reviews=('low_rating', 'sum'),
        low_rating_rate=('low_rating', 'mean')
    )
)

# เอาเฉพาะ category ที่มีรีวิวอย่างน้อย 100 รีวิว
category_summary = (
    category_summary[
        category_summary['total_reviews'] >= 100
    ]
    .sort_values('low_rating_rate', ascending=False)
)

import matplotlib.pyplot as plt

top_categories = category_summary.head(10).sort_values('low_rating_rate')

top_categories['low_rating_rate'].plot(
    kind='barh',
    figsize=(8, 5)
)

plt.xlabel('สัดส่วนรีวิว 1–2 ดาว')
plt.ylabel('Category')
plt.title('สัดส่วนรีวิวคะแนนต่ำในแต่ละหมวดสินค้า')
plt.show()
````

จากกราฟพบว่า Sega Systems, GameCube และ Mac มี <b>สัดส่วนรีวิวคะแนนต่ำ (1–2 ดาว)</b> สูงเมื่อเทียบกับหมวดหมู่อื่น จึงนำไปสู่การวิเคราะห์ต่อว่า รีวิวคะแนนต่ำในหมวดหมู่เหล่านี้กล่าวถึงปัญหาอะไรบ่อยที่สุด และปัญหาดังกล่าวเกี่ยวข้องกับด้านใดของสินค้า เพื่อค้นหา Pain Points ที่ควรนำไปพิจารณาปรับปรุงสินค้า

````python
# กราฟสนับสนุนเพื่อดู sample size และความพร้อมในการเปรียบเทียบ

table_merge['rating_group'] = table_merge['rating'].apply(
    lambda x: 'Low (1-2)' if x <= 2
    else 'High (4-5)' if x >= 4
    else 'Medium (3)'
)

category_df = table_merge.explode('categories')

# เอา Video Games ออก เพราะเป็นหมวดใหญ่ที่ทุกสินค้ามี
category_df = category_df[
    category_df['categories'] != 'Video Games'
]

category_rating = (
    category_df[
        category_df['rating_group'].isin(['Low (1-2)', 'High (4-5)'])
    ]
    .groupby(['categories', 'rating_group'])
    .size()
    .unstack(fill_value=0)
)

# เลือก category ที่มีข้อมูลรวมมากพอ
category_rating['total'] = category_rating.sum(axis=1)

category_rating = (
    category_rating[
        category_rating['total'] >= 100
    ]
    .sort_values('total', ascending=False)
    .head(10)
)

category_rating[['Low (1-2)', 'High (4-5)']].plot(
    kind='barh',
    figsize=(9, 6)
)

plt.xlabel('จำนวนรีวิว')
plt.ylabel('หมวดหมู่')
plt.title('การเปรียบเทียบรีวิวคะแนนต่ำและสูงตามหมวดหมู่')
plt.legend(title='กลุ่มคะแนน')
plt.show()
````

กราฟนี้ใช้แสดงจำนวนรีวิวคะแนนต่ำและสูงในแต่ละหมวดสินค้า เพื่อดูขนาดของข้อมูลและใช้เป็นข้อมูลประกอบในการเลือกกลุ่มสินค้าที่จะนำไปวิเคราะห์ปัญหาจากข้อความรีวิวต่อไป

### 2.4 เตรียมข้อความและสร้าง Text Features

ก่อนนำข้อความไปวิเคราะห์ จำเป็นต้องมีการ **preprocessing** เพื่อทำให้ข้อความอยู่ในรูปที่เหมาะสมกับเทคนิคที่จะใช้ อย่างไรก็ตาม ในงานนี้ไม่จำเป็นต้องทำ preprocessing ทุกขั้นตอน ให้เลือกทำ **เท่าที่จำเป็น** ตามวัตถุประสงค์ของการวิเคราะห์

ตัวอย่างเช่น การเปลี่ยนข้อความเป็นตัวพิมพ์เล็ก (lower case) การนำ URL หรือ HTML ออก การแบ่งข้อความเป็นคำ (tokenization) หรือการตัด stopwords เช่น `the`, `and`, `is` ออก ในกรณีที่คำเหล่านี้ไม่ได้ช่วยในการวิเคราะห์

**ไม่บังคับให้ใช้ stemming หรือ lemmatization** หากไม่มีเหตุผลว่าขั้นตอนดังกล่าวจะช่วยตอบคำถามของทีมได้ดีขึ้น

หลังจากเตรียมข้อความแล้ว ให้สร้าง **Text Features อย่างน้อย 2 ชนิด** เพื่อนำข้อความมาเปลี่ยนให้อยู่ในรูปข้อมูลที่สามารถนำไปสำรวจ เปรียบเทียบ หรือวิเคราะห์ต่อได้

---

#### Text Feature คืออะไร?

Text Feature คือ **ข้อมูลเชิงตัวเลขที่สกัดหรือสร้างขึ้นจากข้อความ** เพื่อแทนลักษณะหรือความหมายบางอย่างของข้อความนั้น

Text Feature อาจอยู่ในรูปของ

* **ค่าตัวเลขค่าเดียว** เช่น จำนวนคำ หรือ sentiment score
* **ตัวแปรหลายตัว** เช่น `price=1`, `quality=0`, `delivery=1`
* **เวกเตอร์หลายมิติ** เช่น TF-IDF vector หรือ Text Embedding

ตัวอย่างเช่น รีวิว

> `"The product is easy to use but the delivery was very slow."`

สามารถเปลี่ยนเป็น feature ได้หลายรูปแบบ

| Feature                    |                 ตัวอย่างค่าที่ได้ |
| -------------------------- | --------------------------------: |
| จำนวนคำในรีวิว             |                                12 |
| Sentiment score            |                             -0.15 |
| กล่าวถึง usability หรือไม่ |                                 1 |
| กล่าวถึง delivery หรือไม่  |                                 1 |
| กล่าวถึง price หรือไม่     |                                 0 |
| TF-IDF vector              |   `[0.00, 0.41, 0.00, 0.72, ...]` |
| Embedding vector           | `[0.21, -0.08, 0.63, ..., -0.14]` |

ดังนั้นข้อความหนึ่งข้อความสามารถถูกแทนได้ทั้งด้วย **feature ที่ตีความได้โดยตรง** และ **feature แบบเวกเตอร์ที่ใช้แทนเนื้อหาหรือความหมายของข้อความ**

---

## ตัวอย่าง Text Features ที่สามารถเลือกใช้

ผมจะแก้ตารางเดิมเป็นแบบนี้:

| Text Feature          | คืออะไร                                     | ตัวอย่าง                     | ใช้ตอบคำถามอะไรได้                                    |
| --------------------- | ------------------------------------------- | ---------------------------- | ----------------------------------------------------- |
| **Review Length**     | จำนวนคำหรือตัวอักษร                         | 120 words                    | รีวิวที่ยาวกว่าได้รับ helpful vote มากกว่าหรือไม่     |
| **Sentiment Score**   | คะแนนความเป็นบวก/ลบ                         | `-0.72`, `+0.85`             | ความรู้สึกลูกค้าสัมพันธ์กับ rating หรือไม่            |
| **Aspect Indicators** | ระบุว่าพูดถึงหัวข้อใด                       | `delivery=1`, `price=0`      | ลูกค้าบ่นเรื่องใดมากที่สุด                            |
| **Lexical Diversity** | ความหลากหลายของคำ                           | unique words / total words   | รีวิวที่เขียนละเอียดมีประโยชน์มากกว่าหรือไม่          |
| **TF-IDF Vector**     | เวกเตอร์ที่ให้น้ำหนักคำเด่นในเอกสาร         | `[0, .35, 0, .81, ...]`      | คำใดแยกกลุ่มรีวิวบวกกับลบ                             |
| **Text Embedding**    | dense vector ที่แทนความหมายโดยรวมของข้อความ | `[.12, -.44, .08, ...]`      | รีวิวใดมีความหมายคล้ายกัน หรือมี customer themes อะไร |
| **Custom Feature**    | feature ที่ทีมสร้างตามโจทย์                 | `refund=1`, `defect_count=3` | ตอบ business question เฉพาะของทีม                     |

### ตัวอย่าง Embedding แบบง่าย

ถ้าสอน `sentence-transformers` มาแล้ว อาจใส่ตัวอย่างสั้น ๆ:

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

texts = [
    "The delivery was very slow.",
    "Shipping took too long.",
    "Excellent sound quality."
]

embeddings = model.encode(texts)

print(embeddings.shape)
```

**รายงานเพียง 3 อย่าง**
1. ทำ preprocessing อะไร และทำไม
2. สร้าง feature อะไรอย่างน้อย 2 ตัว
3. feature เหล่านั้นช่วยตอบคำถามทางธุรกิจอย่างไร

> ไม่ต้องรายงาน vocabulary ก่อน–หลัง เว้นแต่ประเด็นนั้นเกี่ยวข้องกับผลวิเคราะห์

ไม่ใช่การจำแนกประเภทที่ถูกต้อง 100%

````python
# เช็กข้อมูลจาก 2.2 ก่อน
print(table_merge.shape)
print(table_merge.columns.tolist())
````

````python
import re

# ใช้ข้อมูลจาก 2.2 ที่สุ่มและ merge แล้ว
df_2_4 = table_merge.copy()

def clean_text(row):
    # รวม title ของรีวิว + review text
    title = str(row["title_x"]) if pd.notna(row["title_x"]) else ""
    review_text = str(row["text"]) if pd.notna(row["text"]) else ""

    text = f"{title} . {review_text}"

    # lowercase
    text = text.lower()

    # ลบ URL
    text = re.sub(r"http\S+|www\.\S+", " ", text)

    # ลบ HTML tag
    text = re.sub(r"<.*?>", " ", text)

    # ขยาย contraction ที่เกี่ยวกับคำปฏิเสธ
    text = text.replace("n't", " not")

    # ลบช่องว่างซ้ำ
    text = re.sub(r"\s+", " ", text).strip()

    return text

df_2_4["clean_text"] = df_2_4.apply(clean_text, axis=1)

print("ตัวอย่างก่อน-หลัง preprocessing")
print("ก่อน:", df_2_4["text"].iloc[0][:200])
print("หลัง:", df_2_4["clean_text"].iloc[0][:200])

print("\nจำนวนข้อมูลที่ใช้ใน 2.4:", len(df_2_4))
````

````python
# ----------------- 2.4 Text Features -----------------

# Feature 1: ตรวจว่ามีการกล่าวถึงปัญหาหรือ defect หรือไม่
DEFECT_WORDS = [
    "broken",
    "defective",
    "bug",
    "glitch",
    "crash",
    "freeze",
    "lag",
    "malfunction",
    "does not work",
    "doesn't work",
    "not working",
    "stopped working"
]


# Feature 2: Aspect Indicators
# ใช้ระบุว่าปัญหาที่กล่าวถึงเกี่ยวข้องกับเรื่องใด
ASPECTS = {

    "technical": [
        "bug",
        "glitch",
        "crash",
        "freeze",
        "lag",
        "broken",
        "malfunction",
        "not working",
        "does not work"
    ],

    "packaging": [
        "packaging",
        "package",
        "box damaged",
        "damaged box",
        "cracked case",
        "arrived broken"
    ],

    "compatibility": [
        "does not work with",
        "not compatible",
        "incompatible",
        "won't install",
        "will not install",
        "requires update"
    ],

    "price_value": [
        "overpriced",
        "waste of money",
        "not worth",
        "expensive for",
        "too expensive"
    ],

    "online": [
        "server",
        "disconnect",
        "disconnected",
        "matchmaking",
        "connection issue",
        "connection problem",
        "online"
    ],

    "gameplay": [
        "gameplay",
        "story",
        "mission",
        "quest",
        "difficulty",
        "controls"
    ],

    "graphics": [
        "graphics",
        "visual",
        "resolution",
        "fps",
        "frame rate"
    ],

    "controller": [
        "controller",
        "drift",
        "joystick",
        "button"
    ]
}


def has_defect(text):
    return any(keyword in text for keyword in DEFECT_WORDS)


def tag_aspects(text):
    return [
        aspect
        for aspect, keywords in ASPECTS.items()
        if any(keyword in text for keyword in keywords)
    ]


# สร้าง Feature 1
df_2_4["is_defect_review"] = (
    df_2_4["clean_text"]
    .apply(has_defect)
)


# สร้าง Feature 2
df_2_4["aspects"] = (
    df_2_4["clean_text"]
    .apply(tag_aspects)
)


# Feature เพิ่มเติม: จำนวน aspect ที่พบในแต่ละรีวิว
df_2_4["n_aspects"] = (
    df_2_4["aspects"]
    .apply(len)
)


print(
    "จำนวนรีวิวที่มี defect/problem mention:",
    df_2_4["is_defect_review"].sum(),
    f"({df_2_4['is_defect_review'].mean() * 100:.2f}%)"
)


display(
    df_2_4[
        [
            "rating",
            "clean_text",
            "is_defect_review",
            "aspects",
            "n_aspects"
        ]
    ].head(10)
)
````

````python
# ============================================================
# 2.5 หา Insight ที่ตอบ Client Brief
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ใช้ผลลัพธ์จาก 2.4
df_2_5 = final_table.copy()

print("จำนวนรีวิวทั้งหมด:", len(df_2_5))
print("จำนวนรีวิว 1–2 ดาว:", df_2_5["rating"].isin([1, 2]).sum())

print("\nคอลัมน์ที่ใช้ใน 2.5:")
print([
    "rating",
    "categories",
    "aspects",
    "is_defect_review"
])
````

````python
# ============================================================
# Step 1: กลุ่มสินค้าที่มีรีวิวคะแนนต่ำ (1–2 ดาว)
# ============================================================

# เลือกเฉพาะรีวิว 1–2 ดาว
df_low = df_2_5[
    df_2_5["rating"].isin([1, 2])
].copy()

# แยก category ที่เป็น list ออกเป็นแต่ละแถว
df_low_category = df_low.explode(
    "categories"
).dropna(
    subset=["categories"]
).copy()

# นับจำนวนรีวิวต่ำในแต่ละ category
category_low = (
    df_low_category
    .groupby("categories")
    .agg(
        low_rating_count=("rating", "count"),
        avg_rating=("rating", "mean")
    )
    .reset_index()
)

# จำนวนรีวิวต่ำทั้งหมดที่ใช้เป็นฐาน
total_low_reviews = len(df_low)

# คำนวณสัดส่วน
category_low["share_of_low_reviews"] = (
    category_low["low_rating_count"]
    / total_low_reviews
    * 100
)

# เรียงจากจำนวนรีวิวต่ำมากไปน้อย
category_low = (
    category_low
    .sort_values(
        "low_rating_count",
        ascending=False
    )
    .reset_index(drop=True)
)

print("=== กลุ่มสินค้าที่มีรีวิว 1–2 ดาวมากที่สุด ===")

display(
    category_low.head(15)
)
````

````python
# ============================================================
# Step 2: Visualization - กลุ่มสินค้าที่มีรีวิว 1–2 ดาว
# ============================================================

top_categories = category_low.head(10).copy()

top_categories = top_categories.sort_values(
    "low_rating_count"
)

plt.figure(figsize=(10, 6))

bars = plt.barh(
    top_categories["categories"],
    top_categories["low_rating_count"]
)

for bar in bars:
    value = bar.get_width()

    plt.text(
        value + 10,
        bar.get_y() + bar.get_height() / 2,
        f"{int(value):,}",
        va="center"
    )

plt.xlabel("จำนวนรีวิว 1–2 ดาว")
plt.ylabel("กลุ่มสินค้า")
plt.title(
    "Top Product Categories with Low-Rating Reviews (1–2 Stars)"
)

plt.tight_layout()
plt.show()
````

````python
# ============================================================
# Step 3: Pain Points จาก Aspect ของรีวิว 1–2 ดาว
# ============================================================

# เอาเฉพาะรีวิว 1–2 ดาว
df_low_aspect = df_low.copy()

# แยก aspects
df_low_aspect = df_low_aspect.explode(
    "aspects"
).dropna(
    subset=["aspects"]
).copy()

# แยก categories
df_low_aspect = df_low_aspect.explode(
    "categories"
).dropna(
    subset=["categories"]
).copy()

# นับจำนวนการกล่าวถึงแต่ละ aspect ในแต่ละ category
category_aspects = (
    df_low_aspect
    .groupby(
        ["categories", "aspects"]
    )
    .size()
    .reset_index(
        name="mention_count"
    )
)

# เรียงจากปัญหาที่ถูกกล่าวถึงมากที่สุด
category_aspects = (
    category_aspects
    .sort_values(
        ["categories", "mention_count"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

print("=== ปัญหาที่กล่าวถึงในรีวิว 1–2 ดาว แยกตามกลุ่มสินค้า ===")

display(
    category_aspects.head(30)
)
````

### 2.6 Reality Check

ก่อนเอา insight ไปบอกลูกค้า ให้ลองหักล้างมันด้วยตัวเองก่อน อย่างน้อย 1 insight ต้องผ่านขั้นนี้

คำอธิบายทางเลือกที่พบบ่อย
* **ผลของขนาด:** กลุ่มที่ดูโดดเด่นอาจมีตัวอย่างน้อยมาก ตรวจ n ของทุกกลุ่ม
* **ผลของส่วนผสม:** ความต่างอาจมาจากสัดส่วนสินค้าในแต่ละกลุ่มไม่เหมือนกัน ไม่ใช่จากพฤติกรรมลูกค้า ลองคุมให้เทียบกันเฉพาะสินค้าประเภทเดียวกัน
* **ผลของเวลา:** ของเก่ามีเวลาสะสม vote และรีวิวมากกว่า
* **ผลของความยาว:** ข้อความยาวมีโอกาสมีคำที่เราค้นหามากกว่าโดยอัตโนมัติ
* **ผลของคนไม่กี่คน:** ผู้ใช้หรือสินค้าไม่กี่รายอาจครองสัดส่วนใหญ่ของกลุ่ม

เลือกอย่างน้อยหนึ่งคำอธิบาย แล้ว **ทดสอบด้วยตัวเลขจริง** สรุปว่า insight ยังยืนอยู่ได้ หรือต้องปรับถ้อยคำให้แคบลง

````python
# ============================================================
# 2.6 Reality Check
# ตรวจสอบ Insight จาก 2.5 ด้วยผลของขนาด (Size Effect)
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Insight ที่ต้องการตรวจสอบจาก 2.5
# ------------------------------------------------------------
# 1. Video Games - controller = 406
# 2. Video Games - technical = 372
# 3. Accessories - controller = 295
#
# ใน 2.5 ใช้ "จำนวนครั้งที่กล่าวถึง"
# Reality Check จะเปลี่ยนเป็น "สัดส่วนต่อจำนวนรีวิว"
# เพื่อดูว่าอันดับเดิมเกิดจากขนาดของกลุ่มหรือไม่


# ------------------------------------------------------------
# Step 1: สร้าง Review ID
# ------------------------------------------------------------

df_check = df_low.copy()

df_check = df_check.reset_index(drop=False)

df_check = df_check.rename(
    columns={"index": "review_id"}
)


# ------------------------------------------------------------
# Step 2: หาจำนวนรีวิว 1–2 ดาวของแต่ละ Category
# ------------------------------------------------------------

category_review_count = (
    df_check
    .explode("categories")
    .dropna(subset=["categories"])
    .groupby("categories")["review_id"]
    .nunique()
    .reset_index(
        name="low_rating_reviews"
    )
)


# ------------------------------------------------------------
# Step 3: หาจำนวนรีวิวที่กล่าวถึงแต่ละ Aspect
# ------------------------------------------------------------

category_aspect_check = (
    df_check[
        df_check["aspects"].apply(
            lambda x:
            isinstance(x, (list, tuple, np.ndarray))
            and len(x) > 0
        )
    ]
    .explode("aspects")
    .dropna(subset=["aspects"])
    .explode("categories")
    .dropna(subset=["categories"])
)

# นับ "จำนวนรีวิว" ที่กล่าวถึง Aspect
# ไม่ใช่จำนวนครั้งจากการ explode ซ้ำ
category_aspect_count = (
    category_aspect_check
    .groupby(
        ["categories", "aspects"]
    )["review_id"]
    .nunique()
    .reset_index(
        name="mention_reviews"
    )
)


# ------------------------------------------------------------
# Step 4: รวมจำนวนรีวิว + จำนวนที่กล่าวถึง Aspect
# ------------------------------------------------------------

reality_check = category_aspect_count.merge(
    category_review_count,
    on="categories",
    how="left"
)


# ------------------------------------------------------------
# Step 5: คำนวณ Mention Rate
# ------------------------------------------------------------

reality_check["mention_rate"] = (
    reality_check["mention_reviews"]
    / reality_check["low_rating_reviews"]
    * 100
)


# ------------------------------------------------------------
# Step 6: ดูเฉพาะ 3 Pain Points ที่ 2.5 พบว่าสูงสุด
# ------------------------------------------------------------

top3_from_2_5 = pd.DataFrame({
    "categories": [
        "Video Games",
        "Video Games",
        "Accessories"
    ],
    "aspects": [
        "controller",
        "technical",
        "controller"
    ]
})

top3_reality = top3_from_2_5.merge(
    reality_check,
    on=["categories", "aspects"],
    how="left"
)

print("=== Reality Check ของ Top 3 จากข้อ 2.5 ===")

display(
    top3_reality[
        [
            "categories",
            "aspects",
            "mention_reviews",
            "low_rating_reviews",
            "mention_rate"
        ]
    ].round(2)
)


# ------------------------------------------------------------
# Step 7: จัดอันดับใหม่ด้วย Mention Rate
# ------------------------------------------------------------

top3_reality_rank = (
    top3_reality
    .sort_values(
        "mention_rate",
        ascending=False
    )
    .reset_index(drop=True)
)

top3_reality_rank["new_rank"] = (
    range(1, len(top3_reality_rank) + 1)
)

print("\n=== อันดับใหม่หลังควบคุมขนาดของกลุ่ม ===")

display(
    top3_reality_rank[
        [
            "new_rank",
            "categories",
            "aspects",
            "mention_reviews",
            "low_rating_reviews",
            "mention_rate"
        ]
    ].round(2)
)


# ------------------------------------------------------------
# Step 8: ตรวจว่าอันดับเปลี่ยนหรือไม่
# ------------------------------------------------------------

original_order = [
    ("Video Games", "controller"),
    ("Video Games", "technical"),
    ("Accessories", "controller")
]

new_order = list(
    zip(
        top3_reality_rank["categories"],
        top3_reality_rank["aspects"]
    )
)

print("\n=== ผลการ Reality Check ===")

if original_order != new_order:

    print(
        "⚠️ อันดับเปลี่ยนหลังควบคุมจำนวนรีวิว"
    )

    print(
        "\nดังนั้น Insight จากข้อ 2.5 "
        "ไม่สามารถใช้จำนวนครั้งที่กล่าวถึง "
        "เป็นตัวชี้วัดความรุนแรงของปัญหาเพียงอย่างเดียวได้"
    )

    print(
        "ความแตกต่างของจำนวนรีวิว 1–2 ดาว "
        "ในแต่ละ Category มีส่วนทำให้อันดับเดิมแตกต่างกัน"
    )

else:

    print(
        "อันดับยังคงเหมือนเดิมหลังควบคุมจำนวนรีวิว"
    )

    print(
        "ดังนั้น Insight จากข้อ 2.5 "
        "ยังมีหลักฐานสนับสนุนเพิ่มเติม "
        "แม้จะควบคุมผลของขนาดกลุ่มแล้ว"
    )
````

* `GOLD_OIL_KEYWORDS` = คำที่ใช้คัดเฉพาะหัวข้อข่าวทอง/น้ำมัน
* `IMPACT_KEYWORDS` = คำที่ถือว่าหัวข้อข่าว "พูดถึงผลกระทบต่อผู้บริโภค"
* `DELAY` = เวลาพักระหว่างการเรียกแต่ละหน้า เพื่อไม่ยิง request ถี่เกินไป

**1 record ของงานนี้ = 1 หัวข้อข่าว** ที่เก็บจากหน้ารวมข่าวโดยตรง (ไม่ได้เปิดเข้าไปอ่านเนื้อหาบทความ)


````python
# ===== Step 1: ตั้งค่า =====
BASE_URL = "https://www.thaipbs.or.th"
LISTING_URL = "https://www.thaipbs.or.th/news/categories/economy"

# หน้ารวมข่าวหมวดเศรษฐกิจมี 41 หน้า (หน้าสุดท้ายย้อนไปถึงราวต้น เม.ย. 2569)
LISTING_PAGES = [LISTING_URL] + [
    f"{LISTING_URL}/archive?page={i}" for i in range(2, 42)
]

# ใช้ระบุว่าลิงก์ไหนคือ "หัวข้อข่าว" (ไม่ได้เปิดเข้าไปอ่านบทความ)
ARTICLE_PATH_PATTERN = r"^/news/content/\d+$"

# เดือนไทย (ใช้แปลงวันที่ในหน้ารวมข่าวและในหัวข้อข่าว)
THAI_MONTHS = {
    "ม.ค.": 1, "ก.พ.": 2, "มี.ค.": 3, "เม.ย.": 4, "พ.ค.": 5, "มิ.ย.": 6,
    "ก.ค.": 7, "ส.ค.": 8, "ก.ย.": 9, "ต.ค.": 10, "พ.ย.": 11, "ธ.ค.": 12,
}
THAI_MONTH_PATTERN = "|".join(re.escape(m) for m in THAI_MONTHS)

# คำสำคัญไว้กรองว่าหัวข้อข่าวไหนเกี่ยวกับทอง/น้ำมัน
GOLD_OIL_KEYWORDS = ["ทอง", "น้ำมัน", "ดีเซล", "เบนซิน", "แก๊สโซฮอล์", "LPG", "NGV",
                     "กบน", "พลังงาน", "เชื้อเพลิง", "ก๊าซ", "แก๊ส"]

# คำสำคัญไว้เช็คว่า "หัวข้อข่าว" พูดถึงผลกระทบต่อผู้บริโภค/ค่าครองชีพหรือไม่
IMPACT_KEYWORDS = ["ค่าครองชีพ", "ผู้บริโภค", "กำลังซื้อ", "ปากท้อง", "ประชาชน", "ราคาสินค้า",
                   "ของแพง", "ค่าใช้จ่าย", "รายจ่าย", "ต้นทุน", "กระเป๋าเงิน", "ลดภาระ"]

DELAY = 1.5             # วินาที พักระหว่าง request แต่ละครั้ง ไม่ยิงถี่เกินไป
BOT_USER_AGENT = "Part3-CourseProject-Bot/1.0 (educational use)"

session = requests.Session()
session.headers.update({"User-Agent": BOT_USER_AGENT})


def is_gold_oil_news(text):
    return any(kw in text for kw in GOLD_OIL_KEYWORDS)


def mentions_consumer_impact(text):
    return any(kw in text for kw in IMPACT_KEYWORDS)
````

#### Step 2 — ทำความสะอาด URL และตรวจ `robots.txt`

* `normalize_url()` ตัดส่วน `#...` (fragment) และ `/` ท้าย URL ออก เพื่อไม่ให้ URL เดียวกันที่เขียนต่างกันถูกนับซ้ำ
* `robots_allowed()` ตรวจ `robots.txt` ก่อนเข้าแต่ละหน้า ถ้าไม่อนุญาตให้ bot เข้า ต้อง **เปลี่ยนแหล่งข้อมูล** ไม่ใช่หาวิธีหลบ (ถ้าอ่าน `robots.txt` ไม่ได้ จะถือว่าไม่อนุญาตไว้ก่อนเพื่อความปลอดภัย)


````python
# ===== Step 2: normalize URL + ตรวจ robots.txt =====
_robots_cache = {}

def normalize_url(url):
    """ตัด #fragment ออกจาก URL เพื่อไม่ให้หน้าเดียวกันถูกนับซ้ำ"""
    clean_url, _ = urldefrag(url)
    return clean_url.rstrip("/")


def robots_allowed(url, user_agent=BOT_USER_AGENT):
    """คืน True เมื่อ robots.txt อนุญาตให้ bot เข้าถึง URL"""
    p = urlparse(url)
    root = f"{p.scheme}://{p.netloc}"

    if root not in _robots_cache:
        rp = RobotFileParser()
        rp.set_url(urljoin(root, "/robots.txt"))
        try:
            rp.read()
        except Exception:
            # อ่าน robots.txt ไม่ได้ ให้ปลอดภัยไว้ก่อนโดยถือว่าไม่อนุญาต
            _robots_cache[root] = None
            return False
        _robots_cache[root] = rp

    rp = _robots_cache[root]
    if rp is None:
        return False
    return rp.can_fetch(user_agent, url)
````

#### Step 3 — เข้า “หน้ารวมข่าว” เพื่อเก็บหัวข้อข่าวและวันที่

Robot วนเปิดหน้ารวมข่าวทีละหน้า แล้วในแต่ละหน้าทำดังนี้

1. ดู `<a href="...">` ทุกลิงก์ กรองให้เหลือเฉพาะ URL ที่ตรงรูปแบบข่าว (`/news/content/<เลข>`)
2. ดึง **หัวข้อข่าว** จากลิงก์นั้น (`pick_title`)
3. หา **วันที่** จากการ์ดข่าวใบเดียวกัน (`find_date_text`) ซึ่งเว็บแสดง 2 แบบ คือ `"10 ชั่วโมงที่แล้ว"` และ `"27 ก.ย. 69"` แล้วแปลงเป็นวันที่จริง (`parse_listing_date`)
4. เก็บเป็น 1 record ต่อ 1 URL (URL ซ้ำจะถูกเขียนทับ ไม่นับซ้ำ)

ข้อควรรู้: วันที่แบบ "N ชั่วโมงที่แล้ว" จะถูกแปลงโดยอิงเวลาตอนที่รัน จึงควรเก็บข้อมูลแล้วบันทึกเป็น `corpus.jsonl` ไว้ ไม่ต้องรันเก็บใหม่ทุกครั้ง

ขั้นนี้ **ยังไม่ได้เปิดอ่านบทความ** และถ้าหน้าไหนเจอ 0 หัวข้อ โค้ดจะพิมพ์ข้อมูลช่วยดีบั๊ก (⚠️) ออกมาให้


````python
# ===== Step 3: เก็บ "หัวข้อข่าว" + วันที่ จากหน้ารวมข่าวโดยตรง (ไม่เปิดอ่านบทความ) =====

# วันที่ในหน้ารวมข่าวมี 2 แบบ: "10 ชั่วโมงที่แล้ว" (ข่าวใหม่) และ "27 ก.ย. 69" (ข่าวเก่ากว่า)
DATE_TEXT_RE = re.compile(
    rf"(\d+\s*(?:นาที|ชั่วโมง|วัน)ที่แล้ว|\d{{1,2}}\s*(?:{THAI_MONTH_PATTERN})\s*\d{{2,4}})"
)


def clean_text(s):
    return re.sub(r"\s+", " ", s or "").strip()


def parse_listing_date(date_text):
    """แปลงข้อความวันที่ในหน้ารวมข่าวเป็น date เช่น '10 ชั่วโมงที่แล้ว' หรือ '27 ก.ย. 69'"""
    if not date_text:
        return None

    m = re.match(r"(\d+)\s*(นาที|ชั่วโมง|วัน)ที่แล้ว", date_text)
    if m:
        n, unit = int(m.group(1)), m.group(2)
        delta = {"นาที": pd.Timedelta(minutes=n),
                 "ชั่วโมง": pd.Timedelta(hours=n),
                 "วัน": pd.Timedelta(days=n)}[unit]
        return (pd.Timestamp.now(tz="Asia/Bangkok") - delta).date()

    m = re.match(rf"(\d{{1,2}})\s*({THAI_MONTH_PATTERN})\s*(\d{{2,4}})", date_text)
    if m:
        day, month, year = int(m.group(1)), THAI_MONTHS[m.group(2)], int(m.group(3))
        if year < 100:          # ปีแบบย่อ เช่น 69 -> 2569
            year += 2500
        try:
            return pd.Timestamp(year=year - 543, month=month, day=day).date()
        except ValueError:
            return None
    return None


def article_urls_in(node, base_url):
    """คืน set ของ url ข่าวทั้งหมดที่อยู่ใน node นี้"""
    urls = set()
    for x in node.find_all("a", href=True):
        u = normalize_url(urljoin(base_url, x["href"]))
        if re.match(ARTICLE_PATH_PATTERN, urlparse(u).path):
            urls.add(u)
    return urls


def pick_title(anchors):
    """เลือกหัวข้อข่าวจากลิงก์ทั้งหมดที่ชี้ไปข่าวเดียวกัน"""
    # 1) ลิงก์ที่อยู่ใน/ครอบ <h2>,<h3> คือหัวข้อข่าวตัวจริง (ไม่ขึ้นกับว่า a อยู่ใน h3 หรือ h3 อยู่ใน a)
    for a in anchors:
        if a.find_parent(["h2", "h3"]) is not None or a.find(["h2", "h3"]) is not None:
            t = clean_text(a.get_text(" ", strip=True))
            if t:
                return t
    # 2) title attribute ของลิงก์ (เว็บนี้ใส่หัวข้อข่าวไว้ตรงนี้ด้วย แม้ลิงก์ที่เป็นรูปภาพ)
    for a in anchors:
        t = clean_text(a.get("title", ""))
        if t:
            return t
    # 3) ข้อความในลิงก์ที่สั้นที่สุด (ตัวที่ยาวมักมีข้อความเวลาติดมาด้วย)
    texts = [clean_text(a.get_text(" ", strip=True)) for a in anchors]
    texts = [t for t in texts if t]
    return min(texts, key=len) if texts else ""


def find_date_text(anchors, title, url, base_url):
    """หาข้อความวันที่ของการ์ดข่าวใบนี้ โดยไล่ขึ้นไปหา "กล่อง" ที่มีข่าวเดียว (ไม่หยิบของการ์ดข้างเคียง)"""
    for a in anchors:
        node = a
        for _ in range(6):
            node = node.parent
            if node is None or node.name in ("body", "html", "[document]"):
                break
            if article_urls_in(node, base_url) != {url}:
                break       # กล่องนี้มีมากกว่า 1 ข่าว = ไม่ใช่การ์ดใบเดียวแล้ว
            text = clean_text(node.get_text(" ", strip=True)).replace(title, "")
            m = DATE_TEXT_RE.search(text)
            if m:
                return m.group(1)
    return ""


def find_headlines(listing_url):
    """คืน dict {url: record} ของทุกหัวข้อข่าวในหน้ารวมข่าวหน้านี้"""
    if not robots_allowed(listing_url):
        raise PermissionError(f"robots.txt ไม่อนุญาตให้เข้า: {listing_url}")

    r = session.get(listing_url, timeout=20)
    r.raise_for_status()
    soup = BeautifulSoup(r.text, "html.parser")

    # รวบรวมทุก <a> ที่เป็นลิงก์ข่าว แล้วจัดกลุ่มตาม url
    anchors_by_url = {}
    for a in soup.find_all("a", href=True):
        url = normalize_url(urljoin(listing_url, a["href"]))
        if re.match(ARTICLE_PATH_PATTERN, urlparse(url).path):
            anchors_by_url.setdefault(url, []).append(a)

    found = {}
    for url, anchors in anchors_by_url.items():
        title = pick_title(anchors)
        if not title:
            continue
        date_text = find_date_text(anchors, title, url, listing_url)
        found[url] = {
            "url": url,
            "title": title,
            "date_text": date_text,
            "pub_date": parse_listing_date(date_text),
            "category": "เศรษฐกิจ",
        }

    # ถ้าหน้าไหนเจอ 0 หัวข้อ ให้พิมพ์ข้อมูลช่วยดีบั๊กออกมา (พิมพ์ครั้งเดียว)
    if not found and not getattr(find_headlines, "_warned", False):
        find_headlines._warned = True
        print("⚠️ ไม่เจอหัวข้อข่าวในหน้านี้ | <a> ทั้งหมด =", len(soup.find_all("a")),
              "| <h3> =", len(soup.find_all("h3")), "| ความยาว HTML =", len(r.text))
        print("   ตัวอย่างเนื้อหาที่ robot เห็น:", clean_text(soup.get_text(" ", strip=True))[:200])
    return found


all_headlines = {}

for page_url in LISTING_PAGES:
    try:
        items = find_headlines(page_url)
        all_headlines.update(items)
        n_match = sum(1 for v in all_headlines.values() if is_gold_oil_news(v["title"]))
        print(f"{page_url} -> พบ {len(items)} หัวข้อ (สะสมทั้งหมด {len(all_headlines)} | ทอง/น้ำมัน {n_match})")
    except PermissionError as e:
        print("ข้าม (robots.txt ไม่อนุญาต):", e)
    except Exception as e:
        print("ERROR:", page_url, "-", e)
    time.sleep(DELAY)

n_no_date = sum(1 for v in all_headlines.values() if v["pub_date"] is None)
print(f"\nหัวข้อข่าวทั้งหมดที่เก็บได้: {len(all_headlines)} | อ่านวันที่ไม่ได้: {n_no_date}")
````

#### Step 4 — คัดเฉพาะหัวข้อข่าวที่เกี่ยวกับทอง/น้ำมัน

หน้ารวมข่าวหมวดเศรษฐกิจมีข่าวหลายเรื่อง แต่คำถามของทีมสนใจเฉพาะทอง/น้ำมัน จึงคัดหัวข้อที่มีคำใน `GOLD_OIL_KEYWORDS` และเพิ่ม field `mentions_impact` (True/False) ว่าหัวข้อข่าวนั้นมีคำที่พูดถึงผลกระทบต่อผู้บริโภคหรือไม่

หมายเหตุ: `mentions_impact` ดูจากหัวข้อข่าวเท่านั้น ไม่ได้อ่านเนื้อหาเต็ม จึงอาจต่ำกว่าความจริง (ต้องเขียนเป็นข้อจำกัดใน Insight Card)


````python
# ===== Step 4: กรองเฉพาะหัวข้อข่าวที่เกี่ยวกับทอง/น้ำมัน =====
rows = []
for v in all_headlines.values():
    if is_gold_oil_news(v["title"]):
        row = dict(v)
        row["mentions_impact"] = mentions_consumer_impact(v["title"])
        rows.append(row)

print(f"หัวข้อข่าวเศรษฐกิจทั้งหมด {len(all_headlines)} | เกี่ยวกับทอง/น้ำมัน {len(rows)} (ก่อน clean)")
````
